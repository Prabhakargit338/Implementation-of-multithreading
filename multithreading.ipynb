-> A process ia an instance of a computer program.

-> Process runs parallely on  multiple processors >> multiprocessing.

-> Process runs concurrently on multiple threads in a processor >> multithreading.

```python
import time 
start = time.perf_counter()

def test_func():
  print("Do something")
  print("Sleep for 1 second")
  time.sleep(1)
  print("Done with sleeping")
  
test_func()
test_func()
test_func()
test_func()
test_func()
end = time.perf_counter()

print(f"The program finished in {round(end-start, 2)} seconds")
```

Since the program runs sequentially(single thread on a single core/processor) , so it took 5 seconds.

```python
# Now lets say, you want to use multiple threads.
import time 
import threading  # python module
start = time.perf_counter()

def test_func():
  print("Do something")
  print("Sleep for 1 second")
  time.sleep(1)
  print("Done with sleeping")

# Run the program on two threads.
t1 = threading.Thread(target = test_func)
t2 = threading.Thread(target = test_func)

t1.start()  # To start the thread.
t2.start()

t1.join() # join first excute these t1 and t2 threads, then main thread ll be executed .
t2.join()

end = time.perf_counter()

print(f"The program finished in {round(end-start, 2)} seconds")
```

In above the program which was supposed to take 2 seconds, now executed in only 1 second as there is concurrent execution of program.

```python
import time 
import threading  # python module
start = time.perf_counter()

def test_func():
  print("Do something")
  print("Sleep for 1 second")
  time.sleep(1)
  print("Done with sleeping")

threads = []
for i in range(10):
  t = threading.Thread(target = test_func)
  t.start()
  threads.append(t)

for thread in threads:
  thread.join()

end = time.perf_counter()

print(f"The program finished in {round(end-start, 2)} seconds")
```

Since test_func() is called 10 times, it should have taken 10 secs but due to multi threading got completed in 1 seconds

```python
# Using multithreading with function that takes arguements .
import time 
import threading  # python module
start = time.perf_counter()

def test_func(args):
  print("Do something")
  print(f"Sleep for {args} second")
  time.sleep(args)
  print("Done with sleeping")

threads = []
for i in range(10):
  t = threading.Thread(target = test_func, args = [2]) # pass argument in the form of list .
  t.start()
  threads.append(t)

for thread in threads:
  thread.join()

end = time.perf_counter()

print(f"The program finished in {round(end-start, 2)} seconds")
```

```python
# Use-case of multithtreading
# Multithreading works well with I/O bound tasks means where some output has to wait for some input.
# Example : reading/writing files, network communication, databse queries. 
```

```python
# https://github.com/itsfoss/text-script-files
# Downloading files 
import time
import threading
start = time.perf_counter()
url_list = [
  'https://github.com/itsfoss/text-script-files/blob/master/agatha.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/agatha_complete.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/sample_log_file.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/sherlock.txt',
]

data_list = ["data1.txt", "data.txt", "data3.txt", "data4.txt"]

import urllib.request

def file_download(url, filename):
  urllib.request.urlretrieve(url, filename)

threads = []
for i in range(len(url_list)):
  t = threading.Thread(target = file_download, args = [url_list[i],data_list[i]]) 
  t.start()
  threads.append(t)


for thread in threads:
  thread.join()


end = time.perf_counter()


print(f"The program finished in {round(end-start, 2)} seconds")
```

```python
# Multithreading using concurrent.features >> keeps code consise .
import time
import concurrent.futures
start = time.perf_counter()
url_list = [
  'https://github.com/itsfoss/text-script-files/blob/master/agatha.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/agatha_complete.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/sample_log_file.txt',
  'https://github.com/itsfoss/text-script-files/blob/master/sherlock.txt',
]

data_list = ["data1.txt", "data.txt", "data3.txt", "data4.txt"]

import urllib.request

def file_download(url, filename):
  urllib.request.urlretrieve(url, filename)

  with concurrent.futures.ThreadPoolExecutor as executor:
    executor.map(file_download, url_list, data_list) # args >> function-name with arguments of the function .


end = time.perf_counter()


print(f"The program finished in {round(end-start, 2)} seconds")
```

```python
# another use-case
# Shared variable across all the threads 
start = time.perf_counter()
shared_counter = 0
counter_lock = threading.Lock() # Locking the counter for a specific threads

def increment_shared_counter(x):
  global shared_counter # that can be accessed by all the threads
  with counter_lock:
    shared_counter = shared_counter+1
    print(f"Thread {x}: incremented shared counter to {shared_counter}")
    time.sleep(1)

threads = [threading.Thread(target = increment_shared_counter, args=(i,)) for i in [1,2,3,4,5,6]]

for thread in threads :
  thread.start()
for thread in threads :
  thread.join()

end = time.perf_counter()
print(f"The program finished in {round(end-start, 2)} seconds .")

# summary : You have some shared variables across all the threads that is called a shared_counter n it is global and now u want to actually increase the counter .
# There are 6 threads that's why it is taking 6 seconds.
```

```python
# Using concurrent features
# Shared variable across all the threads 
start = time.perf_counter()
shared_counter = 0
counter_lock = threading.Lock() # Locking the counter for a specific threads

def increment_shared_counter(x):
  global shared_counter # that can be accessed by all the threads
  with counter_lock:
    shared_counter = shared_counter+1
    print(f"Thread {x}: incremented shared counter to {shared_counter}")
    time.sleep(1)

with concurrent.futures.ThreadPoolExecutor() as executor :
  thread_args = [1,2,3,4,5,6]
  executor.map(increment_shared_counter, thread_args)

end = time.perf_counter()
print(f"The program finished in {round(end-start, 2)} seconds .")
```

Summary : shared variables can be incremented by individual threads of a process.
